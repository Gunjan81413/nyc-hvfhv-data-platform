```python
daily_gold = spark.table(
    "workspace.default.gold_daily_provider_metrics"
)

print("Daily Provider Gold rows:", daily_gold.count())

print(
    "Daily Provider Gold trips:",
    daily_gold.agg(F.sum("total_trips")).first()[0]
)
```

```python
shared_gold = spark.table(
    "workspace.default.gold_shared_ride_metrics"
)

print(
    "Shared Ride Gold trips:",
    shared_gold.agg(F.sum("total_trips")).first()[0]
)
```
