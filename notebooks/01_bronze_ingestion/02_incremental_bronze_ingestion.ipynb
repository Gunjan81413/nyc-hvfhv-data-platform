```python
FEB_RAW_PATH = (
    "/Volumes/workspace/default/"
    "nyc_hvfhv_data/raw/2026-02/"
)

SOURCE_MONTH = "2026-02"

df_feb_raw = spark.read.parquet(FEB_RAW_PATH)

print("February rows:", df_feb_raw.count())
print("February columns:", len(df_feb_raw.columns))

df_feb_raw.printSchema()
```

```python
from pyspark.sql.functions import min, max

df_feb_raw.select(
    min("pickup_datetime").alias("min_pickup"),
    max("pickup_datetime").alias("max_pickup")
).show(truncate=False)
```

```python
df_bronze = spark.table(
    "workspace.default.bronze_hvfhv_trips"
)

bronze_source_columns = [
    c for c in df_bronze.columns
    if c not in [
        "_source_month",
        "_source_file",
        "_ingested_at"
    ]
]

print(
    "Same source columns:",
    set(df_feb_raw.columns) == set(bronze_source_columns)
)

print(
    "February-only columns:",
    set(df_feb_raw.columns) - set(bronze_source_columns)
)

print(
    "Bronze-only source columns:",
    set(bronze_source_columns) - set(df_feb_raw.columns)
)
```
